```python
from util import Score_calculate as score

# load lable smiles
with open(R'result\lable.txt', 'r') as true_file:
    true_smiles_reloaded = true_file.readlines()

# Read each saved pred_smiles_k* file for evaluation
for k in [0, 3, 5, 10]:
    pred_file_path = rf'\result\Llama3.1-8B-results\clean\pred_smiles_k{k}.txt'
    
    with open(pred_file_path, 'r') as pred_file:
        pred_smiles_reloaded = pred_file.readlines()
    
    # Remove the leading and trailing spaces from each entry
    true_smiles_reloaded = [smile.strip() for smile in true_smiles_reloaded]
    pred_smiles_reloaded = [smile.strip() for smile in pred_smiles_reloaded]

    # Output the evaluation result of the current Kn
    print(f"Evaluating for Kn = {k}...")
    
    # Compute evaluation metrics 
    with score.suppress_rdkit_warnings():
        results = score.evaluate_batch(true_smiles_reloaded, pred_smiles_reloaded)

    score.result_output(results)
```
